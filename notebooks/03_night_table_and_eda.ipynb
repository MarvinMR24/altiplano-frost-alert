# Night table and merged EDA

Loads `data/processed/features.parquet` (built by `pipelines/03_build_night_table.py`) and the tables of `pipelines/03b_eda_merged.py`. Labels come from the same ERA5-Land reanalysis as the features.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Markdown, display

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
P = ROOT / "data" / "processed"
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
```

```python
manifest = json.loads((P / "features_manifest.json").read_text())
print(manifest["night_range"], manifest["rows"], "rows")
pd.DataFrame(manifest["definitions"], index=["definition"]).T
```

```python
df = pd.read_parquet(P / "features.parquet")
df.head()
```

## Class balance by split

```python
pd.read_csv(P / "eda_merged" / "class_balance.csv", index_col=0)
```

```python
pd.read_csv(P / "eda_merged" / "frost_by_cell_split.csv", index_col=0).round(3)
```

```python
fy = pd.read_csv(P / "eda_merged" / "frost_by_year.csv", index_col=0)
ax = fy.plot(marker="o", legend=False, figsize=(9, 4), title="Frost night rate by year (8 cells)")
for s, (lo, hi) in manifest["split_definition"].items():
    ax.axvspan(lo - 0.5, hi + 0.5, alpha=0.08)
    ax.text((lo + hi) / 2, ax.get_ylim()[1], s, ha="center", va="top", fontsize=8)
plt.show()
```

## Drift by split (standardized mean difference vs train)

```python
pd.read_csv(P / "eda_merged" / "drift_std_mean_diff.csv", index_col=0).round(3)
```

## Correlation with the label (training years only)

```python
corr = pd.read_csv(P / "eda_merged" / "corr_with_label_train.csv", index_col=0)
ax = corr.iloc[::-1].plot.barh(legend=False, figsize=(6, 7))
ax.set_xlabel("correlation with frost (train)"); plt.tight_layout(); plt.show()
```

## Sign check and agreement with the AgERA5 night minimum

```python
s = json.loads((P / "eda_merged" / "summary.json").read_text())
print("cloud_cover_prev_day vs sky_emissivity_1800 (train):", round(s["sign_check_corr_cloud_prev_vs_emissivity_train"], 4))
print("label vs AgERA5 agreement:", round(s["label_vs_agera5_agreement_overall"], 4))
pd.read_csv(P / "eda_merged" / "label_vs_agera5.csv", index_col=0).round(3)
```
